# Data Wrangling

```python
import pandas as pd
```

```python
df = pd.read_csv("data/herm_full_edgelist_MODIFIED.csv")
print(f"Dataset dimensions: {df.shape}")
df.head()
```

Check for missing values and duplicates

```python
duplicates = df.duplicated().sum()
missings = df.isnull().sum()

print(f"Duplicates: {duplicates}")
print(f"Missing values: \n{missings}")
```

Dataset has no missing values or duplicates. Next, we do some feature engineering/cleaning (techincally) so that it is easier to work with.

```python
# Encode the types (chemical/electrical) column into binary
df["isElectrical"] = False
df["isElectrical"] = df["isElectrical"] | df["Type"].str.contains(df["Type"].unique()[1])
df.head()
```

```python
# Encode Left vs. Right
df["Source"] = df["Source"].str.strip()
df["Target"] = df["Target"].str.strip()
not_LR = ["PQR", "AVL", "PVR", "RIR"]   # https://www.wormatlas.org/neurons/Individual%20Neurons/Neuronframeset.html

df["isSourceLeft"]  = False
df["isSourceRight"] = False
df["isTargetLeft"]  = False
df["isTargetRight"] = False
for elem in not_LR:
    df["isSourceLeft"]  = df["isSourceLeft"]  | (df["Source"].str.contains(r"^.*[Ll]$") &~ df["Source"].str.contains(elem))
    df["isSourceRight"] = df["isSourceRight"] | (df["Source"].str.contains(r"^.*[Rr]$") &~ df["Source"].str.contains(elem))
    df["isTargetLeft"]  = df["isTargetLeft"]  | (df["Target"].str.contains(r"^.*[Ll]$") &~ df["Target"].str.contains(elem))
    df["isTargetRight"] = df["isTargetRight"] | (df["Target"].str.contains(r"^.*[Rr]$") &~ df["Target"].str.contains(elem))

df.head()
```

```python
# Encode Neuron Motor/Interneuron Types - https://www.wormatlas.org/neurons/Individual%20Neurons/Neuronframeset.html
motor_neuron_patterns = ["^AS[1-11]$", "^AVL$", "^CP[4-7]$", "^D[ABD][1-9].*$", "^DVB$","^HSN[RL]$",
                         "^M[1-5][LR]?$", "^MI$", "^NSM[LR]$", "^PC[A-C][LR]$", "^PD[AB]$", 
                         "^PVN[LR]$", "^PV[VZ]$", "^RIM[LR]$", "^RM[D-H][DV]?[LR]?$", 
                         "^SM[BD][DV][LR]$", "^SPC[LR]$", "^URA[VD][RL]$", "^V[A-D][1-13]$"
                         ]

interneuron_patterns =  ["^ADA[LR]$", "^AI[A-Z][LR]$", "^AV[A-GKL][LR]?$", "^CP[89]$", "^DV[A-F]$",
                         "^DX[13]/[24]$", "^EF[13]/[24]$", "^I[1-6][LR]?$", "^LUA[LR]$", "^MCM[LR]$",
                         "^MI$", "^PDC$", "^PGA$", "^PLN[LR]$", "^PVC[LR]$", "^PV[N-UW-X][LR]?$",
                         "^RI[A-V][LR]?$", "^RMD[DV]?[LR]$", "^RM[F-H][LR]$", "^S[AB][VD][LR]?$",
                         "^SDQ[LR]$", "^S[IM][ABD][VD][LR]$", "^URX[LR]$", "^VA12$", "^VB1$"
                         ]


df["isSourceMotorNeuron"] = False
df["isTargetMotorNeuron"] = False
for elem in motor_neuron_patterns:
    df["isSourceMotorNeuron"] = df["isSourceMotorNeuron"] | df["Source"].str.contains(elem)
    df["isTargetMotorNeuron"] = df["isTargetMotorNeuron"] | df["Target"].str.contains(elem)

df["isSourceInterneuron"] = False
df["isTargetInterneuron"] = False
for elem in interneuron_patterns:
    df["isSourceInterneuron"] = df["isSourceInterneuron"] | df["Source"].str.contains(elem)
    df["isTargetInterneuron"] = df["isTargetInterneuron"] | df["Target"].str.contains(elem)

df.head()
```
