# Data Wrangling

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore") ## suppressing warnings, which isn't best practice but makes the output much more readable
```

```python
df = pd.read_csv("data/herm_full_edgelist_MODIFIED.csv")
print(f"Dataset dimensions: {df.shape}")
df.head()
```

Check for missing values and duplicates

```python
duplicates = df.duplicated().sum()
missings = df.isnull().sum()

print(f"Duplicates: {duplicates}")
print(f"Missing values: \n{missings}")
```

Dataset has no missing values or duplicates. Next, we do some feature engineering/cleaning (techincally) so that it is easier to work with.

```python
# Encode the types (chemical/electrical) column into binary
df["isElectrical"] = False
df["isElectrical"] = df["isElectrical"] | df["Type"].str.contains(df["Type"].unique()[1])
df.head()
```

```python
# Encode Left vs. Right
df["Source"] = df["Source"].str.strip()
df["Target"] = df["Target"].str.strip()
not_LR = ["PQR", "AVL", "PVR", "RIR"]   # https://www.wormatlas.org/neurons/Individual%20Neurons/Neuronframeset.html

df["isSourceLeft"]  = False
df["isSourceRight"] = False
df["isTargetLeft"]  = False
df["isTargetRight"] = False
for elem in not_LR:
    df["isSourceLeft"]  = df["isSourceLeft"]  | (df["Source"].str.contains(r"^.*[Ll]$") &~ df["Source"].str.contains(elem))
    df["isSourceRight"] = df["isSourceRight"] | (df["Source"].str.contains(r"^.*[Rr]$") &~ df["Source"].str.contains(elem))
    df["isTargetLeft"]  = df["isTargetLeft"]  | (df["Target"].str.contains(r"^.*[Ll]$") &~ df["Target"].str.contains(elem))
    df["isTargetRight"] = df["isTargetRight"] | (df["Target"].str.contains(r"^.*[Rr]$") &~ df["Target"].str.contains(elem))

df.head()
```

```python
# Encode Neuron Motor/Interneuron Types - https://www.wormatlas.org/neurons/Individual%20Neurons/Neuronframeset.html
motor_neuron_patterns = ["^AS[1-11]$", "^AVL$", "^CP[4-7]$", "^D[ABD][1-9].*$", "^DVB$","^HSN[RL]$",
                         "^M[1-5][LR]?$", "^MI$", "^NSM[LR]$", "^PC[A-C][LR]$", "^PD[AB]$", 
                         "^PVN[LR]$", "^PV[VZ]$", "^RIM[LR]$", "^RM[D-H][DV]?[LR]?$", 
                         "^SM[BD][DV][LR]$", "^SPC[LR]$", "^URA[VD][RL]$", "^V[A-D][1-13]$"
                         ]

interneuron_patterns =  ["^ADA[LR]$", "^AI[A-Z][LR]$", "^AV[A-GKL][LR]?$", "^CP[89]$", "^DV[A-F]$",
                         "^DX[13]/[24]$", "^EF[13]/[24]$", "^I[1-6][LR]?$", "^LUA[LR]$", "^MCM[LR]$",
                         "^MI$", "^PDC$", "^PGA$", "^PLN[LR]$", "^PVC[LR]$", "^PV[N-UW-X][LR]?$",
                         "^RI[A-V][LR]?$", "^RMD[DV]?[LR]$", "^RM[F-H][LR]$", "^S[AB][VD][LR]?$",
                         "^SDQ[LR]$", "^S[IM][ABD][VD][LR]$", "^URX[LR]$", "^VA12$", "^VB1$"
                         ]


df["isSourceMotorNeuron"] = False
df["isTargetMotorNeuron"] = False
for elem in motor_neuron_patterns:
    df["isSourceMotorNeuron"] = df["isSourceMotorNeuron"] | df["Source"].str.contains(elem)
    df["isTargetMotorNeuron"] = df["isTargetMotorNeuron"] | df["Target"].str.contains(elem)

df["isSourceInterneuron"] = False
df["isTargetInterneuron"] = False
for elem in interneuron_patterns:
    df["isSourceInterneuron"] = df["isSourceInterneuron"] | df["Source"].str.contains(elem)
    df["isTargetInterneuron"] = df["isTargetInterneuron"] | df["Target"].str.contains(elem)

df.head()
```

# Exploratory Data Analysis
Now that the dataset has been feature-engineered with the neuron type & information about neuronal connections, we do EDA to better understand the connectome.

```python
# importing packages for better data viz
import matplotlib.pyplot as plt
import seaborn as sns
```

First, we look at the numeric field, "Weight," which shows the edge weight of each neuronal connection, to see how it looks over the dataset, both with summary statistics and with a histogram.

```python
weight = df['Weight']
weight.agg(
        count="count",
        mean="mean",
        median="median",
        std="std",
        min="min",
        max="max"
)
```

```python
sns.histplot(data=df, x="Weight", bins=140)
plt.xlabel("Edge weight")
plt.show()

sns.histplot(data=df, x="Weight", bins=140)
plt.xlabel("Edge weight")
plt.yscale("log")
plt.show()
```

As we can see, the distribution of neuronal weight is highly positively skewed, with the vast majority of neurons having a weight of < 10, and a few outliers (including the max value, a weight of 142). This does not tell us much about *C. elegans*'s connectome, however, so we continue our EDA by looking at the engineered fields describing the neurons. 

```python
# binary variables
binary_cols = ["isElectrical","isSourceLeft","isSourceRight","isTargetLeft","isTargetRight",
               "isSourceMotorNeuron","isTargetMotorNeuron","isSourceInterneuron","isTargetInterneuron"]

## summary statistics for each binary variable
for col in binary_cols:
    print(df.groupby(col)["Weight"].agg(
        count="count",
        mean="mean",
        median="median",
        std="std",
        min="min",
        max="max"
    ))
```

Looking at the counts of the weights of neurons for these binary categories, we can see that there are an unequal number of neuronal connections in each engineered field (true vs false). However, the other descriptive statistics don't tell much, so we will use histograms to aid in visualization of the fields.

```python
for col in binary_cols:
    plt.figure(figsize=(10,4))

    # histogram overlaying false/true for each binary field
    sns.histplot(
        data=df,
        x=weight,
        hue=col,
        bins=140,
        multiple="layer",
        element="step",
        stat="count",
        common_norm=False
    )

    plt.yscale("log")
    plt.axvline(0, color="black", linestyle="--")
    plt.title(f"Edge weight distribution by {col}")
    plt.xlabel("Edge weight")
    plt.ylabel("Count, log scale")

plt.savefig("hist.png")
plt.show()
```

With these histograms overlaying the True and False conditions for each binary category, we get a much better sense of how the *C. elegans* specimen's connectome's neurons are distributed in terms of neuronal edge weight. For most categories, the distributions follow the same-shaped $\text{Gamma}$ distribution, with the True condition having fewer members than the False condition (as expected from the summary statistics). 

An interesting observation we can draw from this is that far more of the heavily weighted neurons are chemical, as opposed to electrical, and more of the most heavily weighted neurons are interneurons as opposed to motorneurons.

Just for fun, we also will create a heatmap to see if there are any unusual overlapping relationships.

```python
# heatmap of binary fields
sns.heatmap(df[binary_cols].corr(), cmap="coolwarm", center=0, annot=True)
plt.savefig("heatmap.png")
plt.show()
```

From the heatmap, we can see that the most strongly correlated fields that aren't impossibilities (i.e, `isTargetLeft` and `isTargetRight`) are the directional combinations (`isTargetRight` and `isSourceRight`, and corresponding for left), and the relation between directions and inter-neurons (i.e., `isTargetLeft` and `isTargetInterneuron`). Interestingly, the field with the least correlation with any other field is `isElectrical`, which suggests that chemical and electrical connections are relatively evenly distributed regardless of other aspects of the connection. Also, these correlations are not very strong, which might be an artifact of the dataset coming from a single *C. elegans* individual, although we cannot say for certain.
